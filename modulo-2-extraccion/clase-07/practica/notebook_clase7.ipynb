La diferencia es en como se esta filtrando los datos, si tenes una base de datos de 500.000 registros por tirar un numero, si conoces lo que necesitas podes filtrarlo con 'WHERE' y reducir el numero de registros a 50.000 por ejemplo, si filtras directamente con '* FROM' estas llamando a todos los registros que haya en la base de datos, lo cual va a consumir tambien mas recursos de la PC.

```python
import sqlite3
import pandas as pd
 
def buscar_mascotas(especie=None, edad_maxima=None, ciudad=None):
    conexion = sqlite3.connect("vetcare.db")

    join_duenos = ciudad is not None

    if join_duenos:
        base = "SELECT mascotas.* FROM mascotas JOIN duenos ON mascotas.dueno_id = duenos.id"
    else:
        base = "SELECT * FROM mascotas"

    condiciones = []
    parametros  = []

    if especie is not None:
        condiciones.append("mascotas.especie = ?" if join_duenos else "especie = ?")
        parametros.append(especie)

    if edad_maxima is not None:
        condiciones.append("mascotas.edad <= ?" if join_duenos else "edad <= ?")
        parametros.append(edad_maxima)

    if ciudad is not None:
        condiciones.append("duenos.ciudad = ?")
        parametros.append(ciudad)

    if condiciones:
        consultas = base + " WHERE " + " AND ".join(condiciones)
    else:
        consultas = base

    df = pd.read_sql (consultas, conexion, params=parametros)
    conexion.close()
    return df




    
 
# Probar con distintas combinaciones:
print(buscar_mascotas(especie="perro"))
print(buscar_mascotas(edad_maxima=3))
print(buscar_mascotas(especie="gato", ciudad="Buenos Aires"))
print(buscar_mascotas())  # sin filtros → devuelve todas


# Tu tarea: construir la consulta dinámicamente
    # Si especie='perro', filtrar por especie
    # Si edad_maxima=5, filtrar mascotas con edad <= 5
    # Si ciudad='Buenos Aires', hacer JOIN con duenos y filtrar por ciudad
    # Si se pasan varios, combinar todos los filtros con AND
```

```python
import sqlite3
import pandas as pd
 
conexion = sqlite3.connect("vetcare.db")
 
df = pd.read_sql("SELECT veterinario, costo FROM consultas", conexion)
conexion.close()

veterinarios = df["veterinario"].unique()

for nombre in veterinarios:
    consultas_vet = df[df["veterinario"] == nombre]
    cantidad = len(consultas_vet)
    total = sum(consultas_vet['costo'])

    print(f"Veterianio: {nombre}")
    print(f"Cantidad de consultas: {cantidad}")
    print(f"Total facturado: ${total:.2f}\n")

#Necesite unas 8 lineas sin contar los print ni los import, contando todo son 13 lineas en total.

```
