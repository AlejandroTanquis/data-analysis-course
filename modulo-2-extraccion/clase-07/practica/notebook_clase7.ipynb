```python
import sqlite3
import pandas as pd
 
conexion = sqlite3.connect("vetcare.db")

estructura = pd.read_sql(
    "PRAGMA table_info(duenos)",
    conexion
)

estructura2 = pd.read_sql(
    "PRAGMA table_info(mascotas)",
    conexion
)

estructura3 = pd.read_sql(
    "PRAGMA table_info(consultas)",
    conexion
)

print(estructura)
print(estructura2)
print(estructura3)

conexion.close()
```

```python
import sqlite3
import pandas as pd
 
conexion = sqlite3.connect("vetcare.db")

tablas = ["duenos", "mascotas", "consultas"]

for tabla in tablas:
    resultado = pd.read_sql(
        f"SELECT COUNT(*) AS total FROM {tabla}",
        conexion
    )
print(f"{tabla}: {resultado['total'][0]} filas")
```
