```python
productos = pd.DataFrame([
    {"nombre": "Notebook", "precio": 320000, "stock": 5},
    {"nombre": "Mouse",    "precio": 8500,   "stock": 42},
    {"nombre": "Teclado",  "precio": 45000,  "stock": 12},
])
 
resultado_a = productos["precio"]
resultado_b = productos[["precio"]]
resultado_c = productos[["nombre", "precio"]]
 
print(type(resultado_a))
print(type(resultado_b))
print(type(resultado_c))

#> **Para pensar:** `resultado_a` y `resultado_b` muestran información parecida cuando los imprimes, pero son de tipos distintos. 
# ¿Cuál es Series y cuál es DataFrame? ¿Por qué creés que pasa eso?

#Respuesta: Resultado "a" es un dato de tipo Series y resultado "b" es un dato de tipo DataFrame. Esto ocurre porque al poner un solo corchete lo interpreta como
#un dato de tipo Series, al usar doble corchete lo interpreta como un dato de tipo DataFrame. Series es una secuencia de datos y un DataFrame es una tabla.
# Si bien los 2 pueden mostrar los mismos datos, se muestran de manera distinta, son parecidos pero no son iguales.
```
