```python
def calcular_orden(precio_unitario, cantidad, descuento=0):
    subtotal = precio_unitario * cantidad
    monto_descuento = subtotal * descuento
    total_final = subtotal - monto_descuento
    return subtotal, monto_descuento, total_final

Subtotal, descuento, total_final = calcular_orden (12000, 3,)
print (f"subtotal: $ {Subtotal}) | descuento $ {descuento} | total: $ {total_final}")

Subtotal, descuento, total_final = calcular_orden (8500, 5, descuento=0.10)
print (f"subtotal: ${Subtotal} | descuento: ${descuento} | total: ${total_final}")

Subtotal, descuento, total_final = calcular_orden (45000, 2, descuento=0.25)
print (f"subtotal: ${Subtotal} | descuento: ${descuento} | total ${total_final}")

```

```python
def convertir_temperatura(registro):
    try:
        celsius = float(registro["valor"])
        return (celsius * 9/5) + 32
    except ValueError:
        return None
 
sensores = [
    {"id": "S-01", "valor": "87.3"},
    {"id": "S-02", "valor": "N/A"},
    {"id": "S-03", "valor": "92.1"},
    {"id": "S-04", "valor": ""},
    {"id": "S-05", "valor": "88.7"},
]
 
for sensor in sensores:
    fahrenheit = convertir_temperatura(sensor)
    if fahrenheit is not None:
        print(f"{sensor['id']}: {fahrenheit:.1f}°F")
    else:
        print (f"{sensor['id']}: sin lectura")

```

```python
registros = [
    {"producto": "Notebook",  "precio": "320000", "stock": "5",  "categoria": "tech"},
    {"producto": "Escritorio","precio": "85000",  "stock": "N/A","categoria": "muebles"},
    {"producto": "Mouse",     "precio": "-8500",  "stock": "42", "categoria": "tech"},
    {"producto": "",          "precio": "45000",  "stock": "8",  "categoria": "muebles"},
    {"producto": "Monitor",   "precio": "210000", "stock": "0",  "categoria": "tech"},
    {"producto": "Auriculares","precio": "32000", "stock": "11", "categoria": "tech"},
]
 
resultado = []
errores = []
 
for r in registros:
    try:
        if r["producto"] == "":
            raise ValueError("nombre vacío")
        precio = float(r["precio"])
        if precio <= 0:
            raise ValueError(f"precio inválido: {r['precio']}")
        stock = int(r["stock"])
        valor_inventario = precio * stock
        resultado.append({
            "producto": r["producto"],
            "precio": precio,
            "stock": stock,
            "valor": valor_inventario,
            "categoria": r["categoria"],
        })
    except (ValueError, KeyError) as e:
        errores.append({"producto": r.get("producto", "?"), "error": str(e)})
 
tech    = [r for r in resultado if r["categoria"] == "tech"]
muebles = [r for r in resultado if r["categoria"] == "muebles"]
 
print(f"Procesados: {len(resultado)} | Errores: {len(errores)}")
print(f"\nTech    — valor total: ${sum(r['valor'] for r in tech):,}")
print(f"Muebles — valor total: ${sum(r['valor'] for r in muebles):,}")
 
for e in errores:
    print(f"  ✗ {e['producto']}: {e['error']}")
```

```python
def limpiar_registro(r):
    if r["producto"] == "":
        raise ValueError("nombre vacio")

    precio = float(r["precio"])
    if precio <= 0:
        raise ValueError(f"precio invalido: {r['precio']}")

    stock = int(r["stock"])

    return {
        "producto": r["producto"],
        "precio":   precio,
        "stock":    stock,
        "categoria": r["categoria"],
    
    }

def calcular_valor(registro):
    return registro["precio"] * registro["stock"]

def agrupar_por_categoria(registros):
    grupos = {}
    for r in registros:
        cat = r["categoria"]
        if cat not in grupos:
            grupos[cat] = []
        grupos[cat].append(r)
    return grupos

def procesar_dataset(registros):
    procesados = []
    errores = []

    for r in registros:
        try:
            limpio = limpiar_registro(r)
            limpio["valor"] = calcular_valor(limpio)
            procesados.append(limpio)
        except (ValueError, KeyError) as e:
            errores.append ({"producto": r.get("producto", "?"), "error": str(e)})
    return procesados, errores

registros = [
    {"producto": "Notebook",   "precio": "320000", "stock": "5",   "categoria": "tech"},
    {"producto": "Escritorio", "precio": "85000",  "stock": "N/A", "categoria": "muebles"},
    {"producto": "Mouse",      "precio": "-8500",  "stock": "42",  "categoria": "tech"},
    {"producto": "",           "precio": "45000",  "stock": "8",   "categoria": "muebles"},
    {"producto": "Monitor",    "precio": "210000", "stock": "0",   "categoria": "tech"},
    {"producto": "Auriculares","precio": "32000",  "stock": "11",  "categoria": "tech"},
]
 
procesados, errores = procesar_dataset(registros)
grupos = agrupar_por_categoria(procesados)
 
print(f"Procesados: {len(procesados)} | Errores: {len(errores)}")
 
for categoria, items in grupos.items():
    total = sum(r["valor"] for r in items)
    print(f"\n{categoria.capitalize()} — valor total: ${total:,}")
 
for e in errores:
    print(f"  ✗ {e['producto']}: {e['error']}")





```

```python
def limpiar_precio(valor):
    try:
        limpio = valor.strip().replace("$", "").replace(",", "").replace(".", "")
        return float(limpio) and int(limpio)
    except ValueError:
        return None

precios_crudos = ["$45.000", "$1,250.50", " $8500 ", "N/A", "$32.000"]


precios_limpios = [limpiar_precio(p) for p in precios_crudos]
print(precios_limpios)

```
